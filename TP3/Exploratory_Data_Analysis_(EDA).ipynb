```python
import pandas as pd


url_gz = "https://archive.ics.uci.edu/ml/machine-learning-databases/00492/Metro_Interstate_Traffic_Volume.csv.gz"
df = pd.read_csv(url_gz, compression='gzip')

#  Overall Descriptive Statistics for traffic_volume
traffic = df['traffic_volume']


mean_val = traffic.mean()
median_val = traffic.median()
mode_val = traffic.mode()[0]

# حساب التشتت
std_val = traffic.std()
range_val = traffic.max() - traffic.min()


print("📊 Descriptive Statistics for 'traffic_volume':\n")
print(f"Mean (المتوسط): {mean_val:.2f}")
print(f"Median (الوسيط): {median_val:.2f}")
print(f"Mode (المنوال): {mode_val}")
print(f"Standard Deviation (الانحراف المعياري): {std_val:.2f}")
print(f"Range (المدى): {range_val}")

# مقارنة الوسيط بالمتوسط
if mean_val > median_val:
    print("\n🔎 التفسير: المتوسط أكبر من الوسيط → التوزيع مائل نحو اليمين (يوجد ساعات بها حركة مرور عالية جدًا).")
elif mean_val < median_val:
    print("\n🔎 التفسير: المتوسط أصغر من الوسيط → التوزيع مائل نحو اليسار (معظم الساعات بها حركة أقل).")
else:
    print("\n🔎 التفسير: المتوسط ≈ الوسيط → التوزيع متماثل تقريبًا.")
```

The **median** traffic volume is **approximately 3380 vehicles per hour**, while the **mean** traffic volume is around **3260 vehicles per hour**.
The mean is slightly lower than the median, indicating a **left-skewed distribution.**

This suggests that most hours experience **moderate to high traffic**, but there are some hours with very low **traffic volumes** ( late-night periods) that pull the mean downward.
 Overall, the traffic volume data appears to be **fairly balanced**, with only a mild skew toward lower values.

# Calculate the skewness and kurtosis for the traffic_volume column using Pandas or SciPy functions.

```python
import pandas as pd


url_gz = "https://archive.ics.uci.edu/ml/machine-learning-databases/00492/Metro_Interstate_Traffic_Volume.csv.gz"
df = pd.read_csv(url_gz, compression='gzip')

# حساب الالتواء والتفلطح لعمود traffic_volume
skewness = df['traffic_volume'].skew()
kurtosis = df['traffic_volume'].kurtosis()

print(f"📈 Skewness (الالتواء): {skewness:.4f}")
print(f"📊 Kurtosis (التفلطح): {kurtosis:.4f}")
```

**Interpret the sign and magnitude of the skewness value. Does this match your anticipation from Task A?**
The **skewness** value is **-0.0894**, which is very close to **zero** and slightly **negative**.
This indicates that the distribution of **traffic_volume** is **almost symmetrical**, with a small tendency toward **left skewness** (a few lower values pulling the tail slightly to the left).

 This matches our anticipation from **Task A**, where the **mean (≈ 3260)** was slightly lower than the **median (≈ 3380)**, suggesting a mild **left-skewed **distribution.

**Interpret the kurtosis value. Is the distribution heavily-tailed (leptokurtic) or lightly-tailed (platykurtic) compared to a normal distribution?**
The **kurtosis** value is **-1.3091**, which is **less than 0**, indicating a **platykurtic** distribution.
This means the distribution is **flatter** **and more spread out** than a normal distribution, with **fewer extreme outliers and lighter tails**.

# Task C: Multivariate Analysis via Grouping

**Use the groupby() function to calculate the mean and standard deviation of traffic_volume for each unique category in the weather_main column.**

```python
import pandas as pd


url_gz = "https://archive.ics.uci.edu/ml/machine-learning-databases/00492/Metro_Interstate_Traffic_Volume.csv.gz"
df = pd.read_csv(url_gz, compression='gzip')

# حساب المتوسط والانحراف المعياري لحجم المرور حسب حالة الطقس
grouped_stats = df.groupby('weather_main')['traffic_volume'].agg(['mean', 'std']).sort_values(by='mean', ascending=False)


print("📊 Average and Standard Deviation of Traffic Volume by Weather Condition:\n")
print(grouped_stats)
```

**Highest Variability (Standard Deviation):**
 **Fog (std = 2125.53)**
 Traffic volume fluctuates the most during foggy conditions — likely because fog can vary in intensity and timing, sometimes leading to normal activity and other times to severe slowdowns.

** Interpretation:**

**Cloudy conditions** are associated with **the most stable and high traffic flow**, suggesting routine, everyday weather.

**Foggy condition**s cause **inconsistent traffic behavior**, reflecting how unpredictable low-visibility conditions can affect driver decisions.

# Analysis Question (Correlation):

**Which of the environmental variables (temp, rain_1h, snow_1h, clouds_all) has the strongest absolute correlation with traffic_volume? Is this correlation positive or negative?**

Among the environmental variables, **temperature (temp)** has the **strongest absolute correlation** with **traffic_volume (r = 0.13)**.

Although the correlation is weak, it still indicates a mild trend where **warmer conditions are associated with higher road activity**.


# Discussion Question (Causation):



The correlation found in **Task B**  between **temperature** and **traffic volume** — does **not imply causation**.  
Even though higher temperatures are slightly associated with increased traffic, this relationship does **not mean** that warmer weather **causes** more vehicles to be on the road.

According to **Simpson’s Paradox**, aggregated data can hide the influence of other underlying factors that actually drive both variables.  
For example, an **external variable** such as **time of day**, **day of the week**, or **season** could explain the observed correlation:
- During **daytime** or **summer months**, both temperature and traffic volume naturally increase.
- During **nighttime** or **winter**, both tend to decrease.

Thus, the apparent relationship between temperature and traffic volume may simply reflect **a shared pattern influenced by time-related or social factors**, rather than a direct cause-and-effect link.
